## 1. Read and Display an Image Using OpenCV

**Aim:** Read an image named `flower.jpg` using OpenCV and display it in a window titled **Original Image**.

**Explanation:**  `cv2.imread()` reads an image. `cv2.imshow()` displays it in a window with the given title. `cv2.waitKey(0)` waits until a key is pressed, and `cv2.destroyAllWindows()` closes the OpenCV windows.

**Expected Output:** The image is displayed in a window titled **Original Image**.

**Result:** The program performs the stated image-processing operation.

````python
import cv2

# Read the image
img = cv2.imread("flower.jpg")

# Check whether the image was loaded successfully
if img is not None:
    cv2.imshow("Original Image", img)
    cv2.waitKey(0)
    cv2.destroyAllWindows()
else:
    print("Error: Image not found. Check the file path.")
````

## 2. Print Image Dimensions and Data Type

**Aim:** Print the height, width, number of channels and data type of `scene.jpg`.

**Explanation:** `img.shape` returns the image dimensions as height, width and channels for a colour image. `img.dtype` gives the data type used to store pixel values, commonly `uint8`.

## 3. Convert a Colour Image into Grayscale and Save It

**Aim:** Load a colour image, convert it to grayscale and save the result as `output_gray.jpg`.

**Explanation:** `cv2.cvtColor()` converts the colour image from OpenCV's BGR format to grayscale using `cv2.COLOR_BGR2GRAY`. `cv2.imwrite()` saves the resulting image.

**Expected Output:** A grayscale image is displayed and saved as `output_gray.jpg`.

**Result:** The program performs the stated image-processing operation.

````python
import cv2

img = cv2.imread("flower.jpg")

if img is not None:
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    cv2.imwrite("output_gray.jpg", gray)
    cv2.imshow("Grayscale Image", gray)
    cv2.waitKey(0)
    cv2.destroyAllWindows()
    print("Grayscale image saved as output_gray.jpg")
else:
    print("Error: Image not found. Check the file path.")
````

## 5. Resize an Image to 50% of Its Original Width and Height

**Aim:** Resize an image to half of its original width and height using OpenCV.

**Explanation:** `img.shape[:2]` obtains the height and width. Multiplying each dimension by `0.5` calculates the new size. `cv2.resize()` creates the resized image.

**Expected Output:** The resized image has 50% of the original width and 50% of the original height.

**Result:** The program performs the stated image-processing operation.

````python
import cv2

img = cv2.imread("flower.jpg")

if img is not None:
    height, width = img.shape[:2]
    new_width = int(width * 0.5)
    new_height = int(height * 0.5)

    resized = cv2.resize(img, (new_width, new_height))

    cv2.imshow("Original Image", img)
    cv2.imshow("Resized Image (50%)", resized)
    cv2.waitKey(0)
    cv2.destroyAllWindows()
    print("Original dimensions:", width, "x", height)
    print("Resized dimensions:", new_width, "x", new_height)
else:
    print("Error: Image not found. Check the file path.")
````

## 6. Crop the Central 200 × 200 Pixel Area Using Matplotlib

**Aim:** Crop the central 200 × 200 pixel area of an image and display it using Matplotlib.

**Explanation:** `height // 2` and `width // 2` calculate the centre coordinates. NumPy slicing selects 100 pixels on either side of the centre in both directions. BGR is converted to RGB because Matplotlib expects RGB channel order.

**Expected Output:** Matplotlib displays the central 200 × 200 pixel crop.

**Result:** The program performs the stated image-processing operation.

````python
import cv2
import matplotlib.pyplot as plt

img = cv2.imread("flower.jpg")

if img is not None:
    # Convert BGR to RGB for correct Matplotlib colours
    img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    height, width = img.shape[:2]

    if height >= 200 and width >= 200:
        center_x = width // 2
        center_y = height // 2

        cropped = img[
            center_y - 100:center_y + 100,
            center_x - 100:center_x + 100
        ]

        plt.imshow(cropped)
        plt.title("Central 200 x 200 Crop")
        plt.axis("off")
        plt.show()
    else:
        print("Error: Image must be at least 200 x 200 pixels.")
else:
    print("Error: Image not found. Check the file path.")
````

## 9. Convert RGB to HSV and Display Each Channel

**Aim:** Convert an image from RGB to HSV colour space and display the Hue, Saturation and Value channels separately using Matplotlib.

**Explanation:** The image is first converted from BGR to RGB, then from RGB to HSV. `cv2.split()` separates the three channels: Hue represents colour, Saturation represents colour intensity, and Value represents brightness. Matplotlib displays each channel separately in grayscale.

**Expected Output:** Three separate grayscale channel visualisations—Hue, Saturation and Value—are displayed.

**Result:** The program performs the stated image-processing operation.

````python
import cv2
import matplotlib.pyplot as plt

img = cv2.imread("flower.jpg")

if img is not None:
    rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    hsv = cv2.cvtColor(rgb, cv2.COLOR_RGB2HSV)

    hue, saturation, value = cv2.split(hsv)

    plt.figure(figsize=(12, 4))

    plt.subplot(1, 3, 1)
    plt.imshow(hue, cmap="gray")
    plt.title("Hue")
    plt.axis("off")

    plt.subplot(1, 3, 2)
    plt.imshow(saturation, cmap="gray")
    plt.title("Saturation")
    plt.axis("off")

    plt.subplot(1, 3, 3)
    plt.imshow(value, cmap="gray")
    plt.title("Value")
    plt.axis("off")

    plt.tight_layout()
    plt.show()
else:
    print("Error: Image not found. Check the file path.")
````

## 10. Calculate Mean Squared Error (MSE) Between Two Grayscale Images

**Aim:** Compute the Mean Squared Error between `img1.jpg` and `img2.jpg`, both read as grayscale images.

**Explanation:** Both images are loaded in grayscale and checked to ensure they have identical dimensions. Their pixel values are converted to floating-point numbers, squared differences are calculated, and `np.mean()` returns the average squared difference. An MSE of 0 means the images are identical pixel by pixel.

**Expected Output:** The program prints the MSE value. The value depends on the contents of the two input images.

**Result:** The program performs the stated image-processing operation.

````python
import cv2
import numpy as np

img1 = cv2.imread("img1.jpg", cv2.IMREAD_GRAYSCALE)
img2 = cv2.imread("img2.jpg", cv2.IMREAD_GRAYSCALE)

if img1 is not None and img2 is not None:
    if img1.shape == img2.shape:
        # Convert to float to avoid unsigned integer arithmetic issues
        mse = np.mean((img1.astype(np.float64) -
                       img2.astype(np.float64)) ** 2)
        print("Mean Squared Error:", mse)
    else:
        print("Error: Both images must have the same dimensions.")
else:
    print("Error: One or both images could not be loaded.")
````
